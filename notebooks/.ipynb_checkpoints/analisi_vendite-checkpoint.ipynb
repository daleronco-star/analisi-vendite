# Analisi delle vendite
**Obiettivo:** analizzare le vendite di un piccolo negozio attraverso Python, Pandas e Matplotlib.

Il progetto comprende il calcolo dei ricavi, l'analisi delle quantità vendute, il riepilogo dei KPI e la visualizzazione dei risultati.
### 1. Creazione dei dati

```python
# Importazione delle librerie per l'analisi e la visualizzazione dei dati
import pandas as pd 
import matplotlib.pyplot as plt 

# Applica uno stile uniforme alle tabelle senza modificare i dati originali
def mostra_tabella(df_tabella):
    return (
        df_tabella.style
        .hide(axis="index")
        .set_properties(**{
            "text-align": "left",
            "white-space": "nowrap"
        })
        .set_table_styles([
            {"selector": "th", "props": [("text-align", "left")]},
            {"selector": "table", "props": [("width", "100%")]}
        ])
    )

# Creazione di un dizionario con i dati delle vendite
dati = {
    "Prodotto": ["Laptop", "Mouse", "Tastiera", "Laptop", "Mouse", "Monitor"],
    "Quantita": [2, 10, 5, 1, 8, 3],
    "Prezzo": [800, 20, 45, 800, 20, 180]
}

# Conversione dei dati in un DataFrame
df = pd.DataFrame(dati)

# Visualizzazione della tabella iniziale
mostra_tabella(df)
```

### 2. Calcolo dei ricavi
Il ricavo di ogni vendita si ottiene moltiplicando la quantità venduta per il prezzo unitario.

```python
# Calcolo del ricavo per ogni riga
df["Ricavo"] = df["Quantita"] * df["Prezzo"]

# Visualizzazione dei dati aggiornati
mostra_tabella(df)
```

### 3. Raggruppamento per prodotto

```python
# Somma dei ricavi generati da ciascun prodotto
ricavi_per_prodotto = (
    df.groupby("Prodotto", sort=False)["Ricavo"]
    .sum()
    .reset_index()
)

mostra_tabella(ricavi_per_prodotto)
```

### 4. Ordinamento dei ricavi

```python
# Ordina i prodotti dal ricavo più alto al più basso
ricavi_per_prodotto = (
    ricavi_per_prodotto
    .sort_values(by="Ricavo", ascending=False)
    .reset_index(drop=True)
)

mostra_tabella(ricavi_per_prodotto)
```

### 5. Analisi delle quantità vendute

```python
# Calcolo della quantità totale venduta per prodotto
quantita_per_prodotto = (
    df.groupby("Prodotto", sort=False)["Quantita"]
    .sum()
    .reset_index()
)

# Ordinamento dalla quantità maggiore alla minore
quantita_per_prodotto = (
    quantita_per_prodotto
    .sort_values(by="Quantita", ascending=False)
    .reset_index(drop=True)
)

mostra_tabella(quantita_per_prodotto)
```

### 6. Calcolo del ricavo totale

```python
# Somma di tutti i ricavi delle vendite
ricavo_totale = df["Ricavo"].sum()

print(f"Ricavo totale: {ricavo_totale:.2f} €")
```

### 7. Calcolo del ricavo medio per vendita

```python
# Calcolo del ricavo medio per riga di vendita
ricavo_medio = df["Ricavo"].mean()

print(f"Ricavo medio per vendita: {ricavo_medio:.2f} €")
```

### 8. Vendita con il ricavo più alto

```python
# Individua il ricavo massimo tra le singole vendite
ricavo_massimo = df["Ricavo"].max()

print(f"Ricavo massimo in una vendita: {ricavo_massimo:.2f} €")
```

### 9. Vendita con il ricavo più basso

```python
# Individua il ricavo minimo tra le singole vendite
ricavo_minimo = df["Ricavo"].min()

print(f"Ricavo minimo in una vendita: {ricavo_minimo:.2f} €")
```

### 10. Riepilogo dei KPI
I KPI (Key Performance Indicators) sono indicatori che aiutano a sintetizzare i risultati principali dell'analisi.

```python
# Creazione di una tabella con i principali indicatori
riepilogo_kpi = pd.DataFrame({
    "Indicatore": [
        "Ricavo totale",
        "Ricavo medio per vendita",
        "Ricavo massimo per vendita",
        "Ricavo minimo per vendita"
    ],
    "Valore": [
        f'{df["Ricavo"].sum():.2f} €',
        f'{df["Ricavo"].mean():.2f} €',
        f'{df["Ricavo"].max():.2f} €',
        f'{df["Ricavo"].min():.2f} €'
    ]
})

mostra_tabella(riepilogo_kpi)
```

### 11. Visualizzazione dei ricavi per prodotto
Il grafico confronta i ricavi totali generati dai diversi prodotti. I colori passano dal verde per il ricavo maggiore al rosso per quello minore.

```python
# Valori dei ricavi, già ordinati dal maggiore al minore
valori = ricavi_per_prodotto["Ricavo"]

# Scala cromatica: verde per il maggiore, rosso per il minore
colori = ["#41966C", "#A5B85A", "#D9A441", "#D95D5D"]

plt.figure(figsize=(10, 6))

# Creazione del grafico a barre
barre = plt.bar(
    ricavi_per_prodotto["Prodotto"],
    valori,
    color=colori,
    width=0.75,
    edgecolor="white",
    linewidth=1.2
)

# Titolo ed etichette degli assi
plt.title(
    "Ricavi per prodotto",
    fontsize=16,
    fontweight="bold",
    pad=20
)

plt.xlabel("Prodotto", fontsize=11)
plt.ylabel("Ricavi (€)", fontsize=11)

# Mostra il valore esatto sopra ogni barra
plt.bar_label(
    barre,
    fmt="%.0f €",
    padding=5,
    fontsize=10,
    fontweight="bold"
)

# Aggiunge una griglia leggera per facilitare il confronto
plt.grid(axis="y", linestyle="--", alpha=0.25)
plt.gca().set_axisbelow(True)

# Rimuove i bordi superiore e destro per uno stile più pulito
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

# Lascia spazio sopra le etichette dei valori
plt.ylim(0, valori.max() * 1.12)

plt.tight_layout()
plt.show()
```

### 12. Analisi dei risultati

L'analisi delle vendite evidenzia i seguenti risultati:

- **Ricavo totale:** 3.525 €.
- **Prodotto con il ricavo maggiore:** Laptop, con 2.400 €.
- **Prodotto con il ricavo minore:** Tastiera, con 225 €.
- **Prodotto con la quantità venduta maggiore:** Mouse, con 18 unità.

Il Laptop genera il ricavo più alto grazie al suo prezzo unitario elevato, mentre il Mouse è il prodotto venduto in maggior quantità, ma contribuisce meno al ricavo totale rispetto al Laptop.

Questi risultati mostrano l'importanza di analizzare sia le quantità vendute sia i ricavi generati da ciascun prodotto.
